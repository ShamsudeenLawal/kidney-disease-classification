```python
import tensorflow as tf
input_shape = [224, 224, 3]
image_size = input_shape[:-1]

train_dir = "../artifacts/data_ingestion/split/train"

train_ds = tf.keras.utils.image_dataset_from_directory(
    train_dir,
    image_size=image_size,
    shuffle=False,
    seed=42
    )
```

```python
methods = [meth for meth in dir(train_ds) if not meth.startswith("_")]
methods
```

```python
with open("class_labels.txt", "w") as file:
    for label in train_ds.class_names:
        file.write(f"{label}\n")
```

```python
with open("class_labels.txt") as file:
    class_names = file.readlines()

class_names = [name.strip() for name in class_names]
class_names
```
